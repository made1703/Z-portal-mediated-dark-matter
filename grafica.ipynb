```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('/home/maria-fajardo/Documents/micromegas_6.3.0/ZpPortal/datos_gammas1txt', sep=r'\s+')
df.columns = df.columns.str.strip() 

grupos = df.groupby("Gamma[GeV]")


for g_nombre, sub_df in grupos:
    fig, ax = plt.subplots(figsize=(6, 4))
    
    ax.scatter(sub_df["Mdm[GeV]"], sub_df["Vchi_sol"], 
               label=f"Gamma = {g_nombre}", 
               color='darkred', alpha=0.7)
    
    ax.set_xscale("log")
    ax.set_yscale("log")
    
    ax.set_title(f"Resultados para $\Gamma = {g_nombre}$")
    ax.set_xlabel(r"$m_{DM}$ [GeV]")
    ax.set_ylabel(r"$V_\chi$")
    ax.grid(True, which="both", ls=":", alpha=0.6)
    
    plt.tight_layout()
    plt.savefig(f"plot_gamma_{g_nombre}.png")
    plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 5))

colors = {1e-6: 'darkblue', 1e-12: 'red', 1e-18: 'purple'}
marcadores = {1e-6: "o", 1e-12: "s", 1e-18: "^"}
etiquetas = {1e-6: r"$\Gamma_\phi = 10^{-6}$",
             1e-12: r"$\Gamma_\phi = 10^{-12}$",
             1e-18: r"$\Gamma_\phi = 10^{-18}$"}


for g_val, sub_df in df.groupby("Gamma[GeV]"):
    g_key = float(f"{g_val:.1e}") 
    
    if g_key in colors:
        ax.scatter(sub_df["Mdm[GeV]"], sub_df["Vchi_sol"],
                   color=colors[g_key],
                   label=r"$\Gamma_\phi = 10^{" + str(int(np.log10(g_key))) + "}$ GeV",
                   s=20,     
                   alpha=1) 

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_ylim(1e-12,1e1)
ax.set_xlabel(r"$m_{DM}$ [GeV]", fontsize=12)
ax.set_ylabel(r"$V_\chi$", fontsize=12)
ax.legend(loc='best', fontsize=9, frameon=True)
ax.grid(True, which="both", ls=":", alpha=0.5)

plt.tight_layout()
plt.savefig("todas_las_gammas.png", dpi=150)
plt.show()
```
